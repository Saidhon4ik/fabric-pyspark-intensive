## SparkSession

SparkSession is the entry point for working with Apache Spark.
It allows us to create DataFrames, read data and execute Spark SQL.

```python
from pyspark.sql import SparkSession

spark = (
        SparkSession.builder
        .master("local[*]")
        .appName("PySparkPractice")
        .getOrCreate()
)

print(spark.version)
```

```python
data = [
    (1, "John", 25, "New York"),
    (2, "Alice", 30, "London"),
    (3, "Bob", 22, "Tashkent"),
    (4, "Sarah", 28, "Berlin")
]

columns = ["id", "name", "age", "city"]

df = spark.createDataFrame(data,columns)

df.show()
```

```python
import time

start = time.time()
df.count()
print(f"Execution time: {time.time() - start:.3f} seconds")
```

```python
df.select("name","age","city").show()
```

```python
df.filter(df.age > 25).show()
```

```python
df.filter((df.age > 25) & (df.city != "London")).show()
```

```python
from pyspark.sql.functions import col

df = df.withColumn('future_age',col('age') + 5)

df.show()
```

```python
df.drop('future_age')
```

```python
from pyspark.sql.functions import col, when

df.select("name", "age", "city") \
    .withColumn(
        "age_category",
        when(col("age") >= 25, "Adult")
        .otherwise("Young")
    ) \
    .show()
```

```python
from pyspark.sql.functions import col

df.select("name","age") \
        .filter(col('age') > 24) \
        .withColumn('age_next_year',col("age") + 1) \
        .drop('city') \
        .show()


```

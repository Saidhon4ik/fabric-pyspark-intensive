```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = (
       SparkSession.builder
       .master("local[*]")
       .appName("joins")
       .getOrCreate()
)


customers_data = [
    (1, "John"),
    (2, "Alice"),
    (3, "Bob")
]

customers_columns = ["customer_id", "name"]

orders_data = [
    (101, 1, 250),
    (102, 1, 150),
    (103, 2, 400),
    (104, 4, 100)
]

orders_columns = ["order_id", "customer_id", "amount"]



customers = spark.createDataFrame(customers_data, customers_columns)

orders = spark.createDataFrame(orders_data, orders_columns)


customers.show()
orders.show()
```

```python
joined = customers.join(
    orders,
    on=customers.customer_id == orders.customer_id,
    how="inner"
)

joined.show()
```

```python
joined_left = customers.join(
    orders,
    on=customers.customer_id == orders.customer_id,
    how="left"
)

joined_left.show()
```

```python
joined_right = customers.join(
        orders,
        on=customers.customer_id == orders.customer_id,
        how = "right"
)

joined_right.show()
```

```python
joined_full = customers.join(
        orders,
        on=customers.customer_id == orders.customer_id,
        how="full"
)

joined_full.show()
```

```python
orders.groupBy("customer_id")
```

```python
orders.groupBy("customer_id").count().show()
```

```python
from pyspark.sql.functions import sum

orders.groupBy("customer_id").agg(
    sum("amount").alias("total_spent")
).show()
```

```python
from pyspark.sql.functions import avg

orders.groupBy("customer_id").agg(
        avg("amount").alias("average_order")
).show()
```

```python
from pyspark.sql.functions import avg,sum,count

orders.groupBy("customer_id").agg(
        count("order_id").alias("total_orders"),
        sum("amount").alias("total_spent"),
        avg("amount").alias("average_order")
).show()
```

```python
#exercise:
joined = customers.join(
        orders,
        on=customers.customer_id == orders.customer_id,
        how="inner"
)

joined_selected = joined.select(
        col("name"),
        col("amount")
)


joined_selected.groupBy("name").agg(
        sum("amount").alias("total_spent")
).show()
```

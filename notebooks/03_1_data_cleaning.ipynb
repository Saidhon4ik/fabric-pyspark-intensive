```python
# Let's put together only the methods we've learned.

# Using your existing df, create a cleaned DataFrame called df_final:
# 1. Remove rows where age is NULL.
# 2. Remove duplicate records.
# 3. Convert id and age to integer.
# 4. Display the final schema and data.
# Your task: Write the transformations yourself using:
# - dropna()
# - dropDuplicates()
# - withColumn()
# - cast()


from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = (
        SparkSession.builder
        .master("local[*]")
        .appName("DataCleaning")
        .getOrCreate()
)


data = [
    (1, "John", 25, "New York"),
    (2, "Alice", None, "London"),
    (3, "Bob", 22, None),
    (4, None, 28, "Berlin"),
    (5, "Sarah", None, "Tashkent"),
    (1, "John", 25, "New York"),
    (2, "Alice", None, "London")
]
columns = ["id", "name", "age", "city"]


df = spark.createDataFrame(data,columns)

df.show()
```

```python
df_filtered = df.dropna(subset=["age"])
df_filtered.show()
```

```python
df_dedup = df_filtered.dropDuplicates()
```

```python
df_final = df_dedup.withColumn("age",col("age").cast("integer"))
df_final = df_final.withColumn("id",col("id").cast("integer"))
```

```python
df_final.printSchema()
df_final.show()
```

## 1. Create a SparkSession

A `SparkSession` is the main entry point for working with Spark. It lets us create DataFrames, read data, and use Spark SQL.

We use `local[*]` to run Spark on this computer and make the available CPU cores usable by the local Spark application.


```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("SparkFundamentals")
    .getOrCreate()
)

print("Spark version:", spark.version)
print("Application:", spark.sparkContext.appName)
print("Master:", spark.sparkContext.master)
```

## 2. Create a DataFrame

A Spark DataFrame is a distributed collection of data organized into named columns. It is conceptually similar to a table in a relational database.


```python
data = [
    (1, "John", 25, "New York"),
    (2, "Alice", 30, "London"),
    (3, "Bob", 22, "Tashkent"),
    (4, "Sarah", 28, "Berlin")
]

columns = ["id", "name", "age", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

## 3. Inspect the schema

A schema describes the structure of a DataFrame: its column names, data types, and whether each column can contain null values.

Spark can infer a schema from the supplied data, as it did when we created this DataFrame.


```python
df.printSchema()
```

## 4. Inspect DataFrame metadata and row count

These properties and methods help us understand an unfamiliar DataFrame before transforming it.


```python
print("Columns:", df.columns)
print("Data types:", df.dtypes)
print("Total records:", df.count())
```

## 5. Transformations and actions

A **transformation** describes a new DataFrame based on an existing one. Examples include filtering rows and selecting columns.

An **action** asks Spark to produce a result. Examples include `show()` and `count()`.

Spark transformations are generally evaluated lazily: Spark records the operations and waits until an action requires a result.


```python
filtered_df = df.filter(df.age > 25)

# The transformation above defines the filter.
# show() is an action that requests the result.
filtered_df.show()
```

## 6. Inspect an execution plan

`explain()` displays how Spark plans to execute a DataFrame query. It is useful when investigating query behavior and performance.

It does not display the DataFrame's rows.


```python
filtered_df.explain()
```

## Key takeaways

- A SparkSession is the entry point to Spark.
- A DataFrame stores structured data in named columns.
- A schema defines column names, types, and nullability.
- Transformations describe operations; actions request results.
- Lazy evaluation allows Spark to plan and optimize work before execution.

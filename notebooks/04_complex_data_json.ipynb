```python
from pyspark.sql import SparkSession

spark = (
        SparkSession.builder
        .master("local[*]")
        .appName("DataCleaning")
        .getOrCreate()
)
```

```python
df = (
    spark.read
    .option("multiLine", "true")
    .json("../data/users.json")
)

df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import col, explode

df.select(
    col("id"),
    col("name"),
    col("address.city").alias("city"),
    col("address.country").alias("country_name"),
    explode(col("skills")).alias("skill")
).show()
```

```python
df_flat = df.select(
        col("id"),
        col("name"),
        col("age"),
        col("address.city").alias("city"),
        col("address.country").alias("country"),
        explode(col("skills").alias("skill"))
)

df_flat.show()
df_flat.printSchema()
```
